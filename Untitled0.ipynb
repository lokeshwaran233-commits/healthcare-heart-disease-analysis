```python
import pandas as pd

df = pd.read_csv("/content/Dataset .csv")  # update path

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['Latitude', 'Longitude'])
```

```python
df = df[(df['Latitude'] != 0) & (df['Longitude'] != 0)]
```

```python
df[['Latitude', 'Longitude']].describe()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
plt.scatter(df['Longitude'], df['Latitude'], alpha=0.4)
plt.title("Restaurant Geographic Distribution")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.show()
```

```python
import folium

# center map
map_center = [df['Latitude'].mean(), df['Longitude'].mean()]
m = folium.Map(location=map_center, zoom_start=2)

# plot points
for _, row in df.iterrows():
    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=2,
        color='blue',
        fill=True,
        fill_opacity=0.6
    ).add_to(m)

m
```

```python
city_counts = df['City'].value_counts()
print(city_counts.head(10))
```

```python
avg_rating_city = df.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False)
print(avg_rating_city.head(10))
```

```python
price_city = df.groupby('City')['Price range'].mean().sort_values(ascending=False)
print(price_city.head(10))
```

```python
city_stats = df.groupby('City').agg({
    'Restaurant ID': 'count',
    'Aggregate rating': 'mean'
}).rename(columns={'Restaurant ID': 'count'})

print(city_stats.sort_values(by='count', ascending=False).head(10))
```

```python
df['Cuisines'] = df['Cuisines'].str.split(',')
```

```python
df_exploded = df.explode('Cuisines')
```

```python
top_cuisines = df_exploded['Cuisines'].value_counts().head(10)
print(top_cuisines)
```

```python
cuisine_city = df_exploded.groupby(['City', 'Cuisines']).size().reset_index(name='count')

top_cuisine_city = cuisine_city.sort_values(['City', 'count'], ascending=[True, False])

print(top_cuisine_city.head(10))
```

```python
top_cities = df['City'].value_counts().head(5).index

filtered = df_exploded[df_exploded['City'].isin(top_cities)]

top_cuisine_top_cities = filtered.groupby(['City', 'Cuisines']).size().reset_index(name='count')

print(top_cuisine_top_cities.sort_values(['City','count'], ascending=[True, False]).head(15))
```
